```python
import pandas as pd
import numpy as np
```

```python
# Loading the cleaned data set

df = pd.read_csv("data/charging_records_cleaned.csv")

print("Dataset loaded successfully")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
df.info()
```

```python
# Convert date columns into datetime format


df["StartDatetime"] = pd.to_datetime(
    df["StartDatetime"],
    errors="coerce"
)

df["EndDatetime"] = pd.to_datetime(
    df["EndDatetime"],
    errors="coerce"
)

print(df[[
    "StartDatetime",
    "EndDatetime"
]].dtypes)


df["StartDay"] = pd.to_datetime(
    df["StartDay"], errors="coerce"
)

df["EndDay"] = pd.to_datetime(
    df["EndDay"], errors="coerce"
)


df["StartTime"] = pd.to_datetime(
    df["StartTime"],
    format="%H:%M:%S",
    errors="coerce"
).dt.time

df["EndTime"] = pd.to_datetime(
    df["EndTime"],
    format="%H:%M:%S",
    errors="coerce"
).dt.time
```

```python
df.info()
```

```python
# Extract the hour

df["Hour"] = df["StartDatetime"].dt.hour

display(df[["StartDatetime", "Hour"]].head(10))
```

```python
# Extract the day of the week

df["DayOfWeek"] = df["StartDatetime"].dt.dayofweek

display(df[["StartDatetime", "DayOfWeek"]].head(10))
```

```python
# Extract month and year

df["Month"] = df["StartDatetime"].dt.month

df["Year"] = df["StartDatetime"].dt.year

display(
    df[["StartDatetime", "Month", "Year"]].head(10)
)
```

```python
# Create weekday/weekend column-----0 means weekday, 1 means weekend

df["IsWeekend"] = (
    df["DayOfWeek"] >= 5
).astype(int)

display(
    df[["DayOfWeek", "IsWeekend"]].head(10)
)
```

```python
# Create time-of-day categories

df["TimeOfDay"] = pd.cut(
    df["Hour"],
    bins=[-1, 5, 11, 17, 23],
    labels=[
        "Night",
        "Morning",
        "Afternoon",
        "Evening"
    ]
)

display(
    df[["Hour", "TimeOfDay"]].head(15)
)
```

```python
# Checking all newly created columns

display(df.head())

print("Total columns:", len(df.columns))

print("Newly created features:")

print([
    "Hour",
    "DayOfWeek",
    "Month",
    "Year",
    "IsWeekend",
    "TimeOfDay"
])
```

```python
# Save the feature-engineered dataset

df.to_csv(
    "data/charging_records_featured.csv",
    index=False
)

print("Feature-engineered dataset saved successfully")
```
